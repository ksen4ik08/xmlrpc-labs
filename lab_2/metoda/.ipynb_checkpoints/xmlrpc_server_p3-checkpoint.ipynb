```python
from xmlrpc.server import SimpleXMLRPCServer
from xmlrpc.server import SimpleXMLRPCRequestHandler

import xmlrpc.client
import datetime
import pickle
import numpy as np
import pandas as pd


class RequestHandler(SimpleXMLRPCRequestHandler):
    rpc_paths = ('/RPC2',)

server = SimpleXMLRPCServer(("localhost", 8008),
                            requestHandler=RequestHandler)
server.register_introspection_functions()
```

```python
# ---------- Статистика (п.3, п.4) ----------
STATS_URL = "http://localhost:8018"

def add_log(log_line):
    """
    Отправляет событие в stats-сервер.
    Если stats-сервер выключен — не падаем (п.4).
    """
    try:
        stats_server = xmlrpc.client.ServerProxy(STATS_URL)
        stats_server.add_log(log_line)
    except Exception:
        pass  # молча игнорируем — сервер статистики может быть выключен
    return True
```

```python
# ---------- Служебные функции (ЛР1) ----------
def ping():
    add_log('ping')
    return True
server.register_function(ping, 'ping')


def now():
    add_log('now')
    return datetime.datetime.now()
server.register_function(now, 'now')


def show_type(arg):
    add_log('type')
    return (str(arg), str(type(arg)), arg)
server.register_function(show_type, 'type')


def test_sum(a, b):
    add_log('sum')
    return a + b
server.register_function(test_sum, 'sum')


def test_pow(a, b):
    add_log('pow')
    return a ** b
server.register_function(test_pow, 'pow')
```

```python
# ---------- Пункт 1: чёрный список ----------
def _levenshtein(a, b):
    """Расстояние Левенштейна (число правок между строками). Без внешних библиотек."""
    a, b = str(a).lower(), str(b).lower()
    if a == b:
        return 0
    if len(a) < len(b):
        a, b = b, a
    prev = list(range(len(b) + 1))
    for i, ca in enumerate(a, 1):
        cur = [i]
        for j, cb in enumerate(b, 1):
            cost = 0 if ca == cb else 1
            cur.append(min(prev[j] + 1, cur[j-1] + 1, prev[j-1] + cost))
        prev = cur
    return prev[-1]


def black_list_check(surname, name, patr, birth_date):
    try:
        frame = pd.read_csv('bad_boys2.csv', header=0, sep=',', encoding='utf-8')

        exact = None
        similar = []

        for _, row in frame.iterrows():
            # все сравнения регистронезависимы (внутри _levenshtein .lower())
            d_s = _levenshtein(row['Surname'],  surname)
            d_n = _levenshtein(row['Name'],     name)
            d_p = _levenshtein(row['Patronym'], patr)
            # дата — точное сравнение
            d_b = 0 if str(row['Birth']).strip() == str(birth_date).strip() else 1

            # полное совпадение ФИО и даты
            if d_s == 0 and d_n == 0 and d_p == 0 and d_b == 0:
                exact = (row['Surname'], row['Name'], row['Patronym'], row['Birth'])
                break

            # частичное: каждое поле ФИО отличается не более чем на 1 букву + дата совпала
            if d_s <= 1 and d_n <= 1 and d_p <= 1 and d_b == 0:
                similar.append(
                    f"{row['Surname']} {row['Name']} {row['Patronym']} {row['Birth']}"
                )

        if exact:
            result = f"{surname} {name} {patr} {birth_date} : yes"
        elif similar:
            result = f"{similar} : similar"
        else:
            result = f"{surname} {name} {patr} {birth_date} : no"

        add_log('black_list_check')
        return result

    except FileNotFoundError:
        return "Error: 'bad_boys2.csv' not found."
    except Exception as e:
        return f"An error occurred: {e}"

server.register_function(black_list_check, 'black_list_check')
```

```python
# ---------- ЛР1: инверсия ----------
def send_back_inversion(bin_data):
    add_log('color_inversion')
    img_arr = pickle.loads(bin_data.data)

    if img_arr.ndim == 3:
        height, width, channels = img_arr.shape
        for i in range(height):
            for j in range(width):
                for k in range(channels):
                    img_arr[i, j, k] = 255 - img_arr[i, j, k]
    elif img_arr.ndim == 2:
        height, width = img_arr.shape
        for i in range(height):
            for j in range(width):
                img_arr[i, j] = 255 - img_arr[i, j]

    return xmlrpc.client.Binary(pickle.dumps(img_arr))
server.register_function(send_back_inversion, 'color_inversion')
```

```python
# ---------- ЛР1: бинаризация ----------
def binarize_image(bin_data, threshold):
    add_log('binarize_image')
    img_arr = pickle.loads(bin_data.data)

    if img_arr.ndim == 3:
        height, width, channels = img_arr.shape
        for i in range(height):
            for j in range(width):
                pixel_value = (int(img_arr[i, j, 0]) +
                               int(img_arr[i, j, 1]) +
                               int(img_arr[i, j, 2])) / 3
                new_val = 255 if pixel_value > threshold else 0
                img_arr[i, j, 0] = new_val
                img_arr[i, j, 1] = new_val
                img_arr[i, j, 2] = new_val
    elif img_arr.ndim == 2:
        height, width = img_arr.shape
        for i in range(height):
            for j in range(width):
                pixel_value = img_arr[i, j]
                img_arr[i, j] = 255 if pixel_value > threshold else 0

    return xmlrpc.client.Binary(pickle.dumps(img_arr))
server.register_function(binarize_image, 'binarize_image')
```

```python
# ---------- Пункт 2: новая функция «процент облачности» ----------
def cloud_percent(bin_data, threshold):
    """
    П.2: бинаризация + возврат ПРОЦЕНТА пикселей выше порога.
    «Оценка процента облачности изображения».
    """
    add_log('cloud_percent')
    img_arr = pickle.loads(bin_data.data)

    if img_arr.ndim == 3:
        gray = img_arr.mean(axis=2)   # цветное — усредняем по каналам
    else:
        gray = img_arr                # монохромное

    total = gray.size
    above = int(np.sum(gray > threshold))
    return float(100.0 * above / total)

server.register_function(cloud_percent, 'cloud_percent')


# ---------- ЛР1: разворот по вертикали ----------
def flip_image_vertical(bin_data):
    add_log('flip_vertical')
    img_arr = pickle.loads(bin_data.data)
    flipped_arr = np.zeros_like(img_arr)

    if img_arr.ndim == 3:
        height, width, channels = img_arr.shape
        for i in range(height):
            for j in range(width):
                flipped_arr[i, j, :] = img_arr[i, width - 1 - j, :]
    elif img_arr.ndim == 2:
        height, width = img_arr.shape
        for i in range(height):
            for j in range(width):
                flipped_arr[i, j] = img_arr[i, width - 1 - j]

    return xmlrpc.client.Binary(pickle.dumps(flipped_arr))
server.register_function(flip_image_vertical, 'flip_vertical')


print("Listening on port 8008...")
server.serve_forever()
```
